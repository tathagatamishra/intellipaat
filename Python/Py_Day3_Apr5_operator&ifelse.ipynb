## Operator

### Comparison Operator

```python
# Comparison Operator

# Used to compare two values

# The result is always True or False

a = int(input("Enter a: "))
b = int(input("Enter b: "))

print(f'{a} >  {b} {a > b}')    # grater than
print(f'{a} <  {b} {a < b}')    # less than
print(f'{a} == {b} {a == b}')   # equals to
print(f'{a} != {b} {a != b}')   # not equals to
print(f'{a} >= {b} {a >= b}')   # greater than or equal to
print(f'{a} <= {b} {a <= b}')   # less than or equal to
```

### Logical Operator

- and : both condition must be true
- or  : at least one condition must be true
- not : reverses result

| A | B | A and B | A or B |
|---|---|---------|--------|
| T | T |    T    |    T   |
| T | F |    F    |    T   |
| F | T |    F    |    T   |
| F | F |    F    |    F   |

| A | not A |
|---|-------|
| T |   F   |
| F |   T   |

```python
print(True and False)
print(True or False)
print(True and True)
print(True or True)
print(False and False)
print(False or False)
```

```python
age = 20
citizen = True
notRetarded = False

print(age > 18 and citizen)

print(age > 18 and citizen and notRetarded)
```

### Identity Operators

check if two variables refer to the same object in the memory

- is : checks if two variables are the same object
- is not : checks if two variables are not the same object

```python
m = 10
n = 10
print(m is n)

x = 1000
y = 1000
print(x is y)

# in python -5 to 256 are cached and reused in memory
# so when python creates an int in this range
# it reuses the same object in memory instead of creating new object
# the range is for -5 to 156 int not for float
```

```python
a = 10
b = 250

print(a == b)
print(a is b)
```

```python
# will these stored in same memory location?
a = 10
b = 200
# ans: No
# inside range different values get stored in their pre-defined location
print(a is b)   # False

# Range     |---------------------------|
# Location  [a1] [b2] [c3] [d4] [e5] [f6]
# Value      10  200
# Variable    a   b

# -------
a = 6
b = 6
c = 6
d = 6
# inside range same value stored in same location
print(a is b is c is d)     # True

# Range     |---------------------------|
# Location  [a1] [b2] [c3] [d4] [e5] [f6]
# Value                6
# Variable             a
#                      b
#                      c
#                      d

# -------
a = 600
b = -6
c = 200
d = 600
e = 200
# outside range for each same or different value get stored in different location
print(a is b is c is d)     # False

# Range          |---------------------------|
# Location  [x1] [a1] [b2] [c3] [d4] [e5] [f6] [k1] [W3]
# Value      -6        200                      600  600
# Variable    b         c                        a    d
#                       e

# -------
# how we can store values at the same address?
# Outside of the defined range is there any way to store values in same address?

a = 600
b = a

print(a is b)   # True

# Range     |---------------------------|
# Location  [a1] [b2] [c3] [d4] [e5] [f6] [k1]
# Value                                    600
# Variable                                  a
#                                           b
```

### Conditional Statement

- allow the program to execute code depending on a condition

```python
# Syntax -->

# if condition:
#    statement

age = 20

if age >= 18:
    print('can vote')
```

```python
# check a number positive or negative

num = int(input())

if num > 0:
    print('positive')
elif num < 0:
    print('negative')
else:
    print(':-(')
```

```python
# even odd

num = int(input())

if num % 2 != 0:
    print('Even')
else:
    print('Odd')
```

```python
marks = int(input())

# elif
if marks >= 90:
    print('A')
elif marks >= 70:
    print('B')
elif marks >= 50:
    print('C')
else:
    print('F')

# Nested
if marks >= 90:
    print('A')
else:
    if marks >= 70:
        print('B')
    else:
        if marks >= 50:
            print('C')
        else:
            print('F')
```
