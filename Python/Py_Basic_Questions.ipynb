```python
# QUESTION-3
# Given the list:
# li = [10, 20, 10, 30, 40, 20, 50, 30]

# Perform the following operations:

# 1. Remove duplicate elements manually (do not use set).
# 2. Remove all elements divisible by 10.
# 3. Sort the final list in descending order.
# 4. Print the final list.

# Note:
# - Use loops and conditions

li = [10, 24, 17, 30, 40, 24, 50, 30]

seen = {}
for item in li:
    seen[item] = True       
unique = list(seen.keys())

filtered = [item for item in unique if item % 10 != 0]

filtered.sort(reverse=True)

print(filtered)
```
