## Line Plot

```python
# whenever we want to compare some value with time then we can use line chart
```

```python
sales = [500, 600, 650, 550, 600, 750]
days = ['M', 'T', "W", 'TH', 'F', 'S']

# plt.plot()
plt.figure(figsize= (6,3)) # length and height of chart
plt.plot(days, sales)
plt.xlabel("Days of week")
plt.ylabel("Sales amount")
plt.title("Weekly ice-cream sales")
plt.grid()
plt.show()
```

```python
sales = [500, 600, 650, 550, 600, 750]
days = ['M', 'T', "W", 'TH', 'F', 'S']

# plt.plot()
plt.figure(figsize= (12,3)) # length and height of chart
plt.plot(days, sales, marker = 'o', color = 'green', markerfacecolor = 'red', linestyle= "-.", linewidth = 5)
plt.xlabel("Days of week")
plt.ylabel("Sales amount")
plt.title("Weekly ice-cream sales")
plt.grid()
plt.show()
```

```python
# Create a variable as sem_exams = [ sem1, sem2,...]
# Create a variable as marks = [90, 87,...]

# Then create a line chart to plot this
```

```python
semester_exam = ['sem1', 'sem2', 'sem3', 'sem4', 'sem5', 'sem6', 'sem7', 'sem8']
student_marks1 = [90, 87, 92, 66, 95, 90, 88, 97]
student_marks2 = [87, 67, 82, 96, 75, 80, 98, 77]

plt.figure(figsize = (12,3)) #length and height
plt.plot(semester_exam, student_marks1, marker = 'o', color = 'blue', markerfacecolor = 'red', linestyle = "-.", linewidth = 2)
plt.plot(semester_exam, student_marks2, marker = '+', color = 'orange', markerfacecolor = 'red', linestyle = "-.", linewidth = 2)
plt.grid()
plt.xlabel("Semesters for the Course")
plt.ylabel("Exam Scores")
plt.title("Scores for the Computer Science Course by Semester")
plt.show()
```

```python
semester_exam = ['sem1', 'sem2', 'sem3', 'sem4', 'sem5', 'sem6', 'sem7', 'sem8']
student_marks1 = [90, 87, 92, 66, 95, 90, 88, 97]
student_marks2 = [87, 67, 82, 96, 75, 80, 98, 77]

plt.figure(figsize = (12,3)) #length and height
plt.plot(semester_exam, student_marks1, label = "S1")
plt.plot(semester_exam, student_marks2, label = 'S2' )
plt.grid()
plt.xlabel("Semesters for the Course")
plt.ylabel("Exam Scores")
plt.title("Scores for the Computer Science Course by Semester")
plt.legend()
plt.show()
```

```python
semester_exam = ['sem1', 'sem2', 'sem3', 'sem4', 'sem5', 'sem6', 'sem7', 'sem8']
student_marks1 = [90, 87, 92, 66, 95, 90, 88, 97]
# student_marks2 = [87, 67, 82, 96, 75, 80, 98, 77]

plt.figure(figsize = (5,3)) #length and height
plt.plot(semester_exam, student_marks1)

plt.xlabel("Semesters for the Course")
plt.ylabel("Exam Scores")
plt.title("Scores for the Computer Science Course by Semester")
plt.yticks(range(20, 150, 20))
plt.xticks(rotation = 45)
plt.grid()
plt.show()
```

```python
Month=['Jan','Feb','Mar','Apr']
Quality=[90,80,70,79]
Error=[5,7,9,6]
plt.figure(figsize=(10,3))
plt.plot(Month,Quality,marker='o', color='blue',markerfacecolor='violet',label="S1" )
plt.plot(Month,Error,marker='>', color='yellow',markerfacecolor='violet',label="S2")

plt.xlabel("Month")
plt.ylabel("Quality")
plt.yticks(range(10,120,10))
plt.grid()
plt.legend(loc="upper left")# "best" for bydefault
plt.show()
```

```python
splits_km = ['0-5', '5-10', '10-15', '15-20', '20-21.1']
# Runner A & B: Pace
runner_a_pace = [5.20, 5.45, 6.10, 5.05, 4.55]
runner_b_pace = [5.00, 5.40, 6.35, 5.20, 5.10]
plt.plot(splits_km, runner_a_pace, marker='o', color='purple',
         markerfacecolor='yellow', linestyle='-', linewidth=2, label='Runner A')
plt.plot(splits_km, runner_b_pace, marker='s', color='teal',
         markerfacecolor='orange', linestyle='--', linewidth=2, label='Runner B')
plt.title("Half Marathon Pace per 5 km Split (Hilly Course)")
plt.xlabel("Split (km)")
plt.ylabel("Pace (min/km - lower is faster)")
plt.gca().invert_yaxis()  # faster pace appears higher on the chart
plt.legend()
plt.grid()
plt.show()
```

```python
Employee_year_exp = [ 3, 5, 10, 12, 15, 18]
Employee_company_salary_Lakhs = [5, 7 , 14, 16, 18, 22]
Employee_Market_salary_lakhs = [6, 8.5, 16, 18, 21, 26]

plt.plot(Employee_year_exp,Employee_company_salary_Lakhs, label = 'Company', marker = 'o', color ='b', markerfacecolor = 'g', linestyle= '-.', linewidth=3)
plt.plot(Employee_year_exp,Employee_Market_salary_lakhs, label = 'Market', marker='*')
plt.title("Employee Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary in Lakhs")
plt.legend()
plt.grid()
plt.show()
```

```python
time_period = ['2019', '2020', '2021', '2022', '2023', '2024']
sales = [120, 90, 150, 200, 180, 250]
profit = [20, 10, 40, 60, 55, 80]

plt.figure(figsize=(12,3))

plt.plot(time_period, sales,
         marker='o',
         linestyle='--',
         linewidth=3,
         label='Sales')

plt.plot(time_period, profit,
         marker='s',
         linestyle='-',
         linewidth=3,
         label='Profit')

plt.title("Company Growth Report", fontsize=16)

plt.xlabel("Year", fontsize=12)
plt.ylabel("Amount", fontsize=12)

plt.grid()
plt.legend(loc='upper left')
plt.show()
```

```python
Qtrs = ['Q1', 'Q2', 'Q3', 'Q4']

sales_2023 = [40, 50, 72, 66]
sales_2024 = [57, 67, 82, 76]
sales_2025 = [60, 77, 92, 96]

plt.figure(figsize=(5, 3))

plt.plot(Qtrs, sales_2023, label="2023")
plt.plot(Qtrs, sales_2024, label="2024")
plt.plot(Qtrs, sales_2025, label="2025")

plt.xlabel("Quarters")
plt.ylabel("Sales")
plt.title("3 Year Sales Comparison")

plt.yticks(range(20, 150, 20))
plt.xticks(rotation=45)

plt.grid()
plt.legend()
plt.show()
```

## Bar chart

```python
# bivariate
# numerical and categorical

prod = ['choco', 'mango', 'pista', 'kulfi']  # -- categorical
sales = [1000, 750, 500, 900] # -- numerical data

plt.bar(prod, sales)
```

```python
# bivariate
# numerical and categorical

prod = ['choco', 'mango', 'pista', 'kulfi']  # -- categorical
sales = [1000, 750, 500, 900] # -- numerical data

plt.figure(figsize = (8, 8))
plt.bar(prod, sales, color = 'yellow')
plt.xlabel("Ice-cream flavours")
plt.ylabel("Sales")
plt.title("Comparing sales of ice -cream flavors")
plt.grid()
plt.show()
```

```python
prod = ['choco', 'mango', 'pista', 'kulfi']  # -- categorical
sales = [1000, 750, 500, 900] # -- numerical data

plt.figure(figsize = (5, 5))
plt.bar(prod, sales, color = ['brown','yellow','green', 'white'])
plt.xlabel("Ice-cream flavours")
plt.ylabel("Sales")
plt.title("Comparing sales of ice -cream flavors")
plt.grid()
plt.show()
```

```python
prod = ['choco', 'mango', 'pista', 'kulfi']  # -- categorical
sales = [1000, 750, 500, 900] # -- numerical data

plt.figure(figsize = (5, 5))
plt.bar(prod, sales, color = ['brown','yellow','green', 'white'], bottom = 100)
plt.xlabel("Ice-cream flavours")
plt.ylabel("Sales")
plt.title("Comparing sales of ice -cream flavors")
plt.grid()
plt.show()
```

```python
prod = ['choco', 'mango', 'pista', 'kulfi']  # -- categorical
sales = [1000, 750, 500, 900] # -- numerical data

plt.figure(figsize = (5, 5))
plt.bar(prod, sales, color = ['brown','yellow','green', 'white'], align='edge')
plt.xlabel("Ice-cream flavours")
plt.ylabel("Sales")
plt.title("Comparing sales of ice -cream flavors")
plt.grid()
plt.show()
```

```python
prod=["Choco","Mango","Kulfi","Pista"]
sales=[750,500,600,800]

plt.figure(figsize=(5,5))
plt.bar(prod,sales, color=["brown","Yellow","Blue","orange"],bottom=100)
plt.xlabel("Icecream falvor")
plt.ylabel("Sales")
plt.show()
```

```python
# Movie data
movies = ["Beserk", "Death Note", "Attack on Titan", "Ghost in the Shell"]
ratings = [9.9, 7.0, 8.0, 8.4]

plt.figure(figsize=(8, 5))

plt.bar(movies, ratings)

plt.xlabel("Anime")
plt.ylabel("Ratings")
plt.title("Anime Rating Comparison")

plt.grid(axis='y')

plt.show()
```

## Pie Chart

```python
# when we want to compare the proportions of the categories
```

```python
fruits = ['apple', 'mango', 'kiwi']
units = [30, 45, 25]

plt.pie(units, labels = fruits)
plt.title("Proportion of fruits on the seller's shop")
plt.show()
```

```python
fruits = ['apple', 'mango', 'kiwi']
units = [30, 45, 25]

plt.pie(units, labels = fruits, autopct = "%0.2f%%")
plt.title("Proportion of fruits on the seller's shop")
plt.show()
```

```python
traffic_type = ['Social_Media', 'Billboard', "Magazine", "Friend"]
contribution = [80, 50, 30, 50]

plt.pie(contribution, labels = traffic_type, autopct= "%0.1f%%", explode = [0,0,0.2,0])
plt.title("Website Traffic from multiple sources")
plt.show()
```
