## Is the Distribution of Sugar Price in US is significantly different from Normal Distribution

```python
print(df['sugar_us'].skew())

plt.figure()
sns.histplot(df['sugar_us'], kde=True)

plt.title("US Sugar Prices Distribution")
plt.xlabel("US Sugar Price")
plt.ylabel("Frequency")

plt.show()
```

```python
from scipy import stats
```

```python
df.columns
```

```python
us_sugar = df['sugar_us']
test_result, p_value = stats.normaltest(us_sugar)
```

```python
p_value
```

```python
if(p_value < 0.05):
  print("Reject the Null Hypothesis")
else:
  print("Failed to Reject the Null Hypothesis")
```

## What is the overall trend of all the commodities?

commodities = ['oil_brent', 'oil_dubai', 'coffee_arabica', 'coffee_robustas', 'tea_columbo', 'tea_kolkata', 'tea_mombasa', 'sugar_eu', 'sugar_us', 'sugar_world']

```python
commodities = [
    'oil_brent', 'oil_dubai',
    'coffee_arabica', 'coffee_robustas',
    'tea_columbo', 'tea_kolkata', 'tea_mombasa',
    'sugar_eu', 'sugar_us', 'sugar_world'
]

plt.figure(figsize=(15, 7))

for commodity in commodities:
    plt.plot(df['date'], df[commodity], label=commodity)

plt.title("Trend of Commodity Prices Over Time")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(5, 2, figsize=(15, 18))
axes = axes.flatten()

for i, commodity in enumerate(commodities):
    axes[i].plot(df['date'], df[commodity])
    axes[i].set_title(commodity)
    axes[i].grid()

plt.tight_layout()
plt.show()
```
