## What is the Skewness of the price distribution of Arabica Coffee?
## Plot the Price Distribution using Histogram.

```python
# Using Pandas

df['coffee_arabica'].skew()
```

```python
# Using SciPy

from scipy.stats import skew

skew(df['coffee_arabica'])
```

```python
# using matplotlib

plt.hist(df['coffee_arabica'], bins=40, edgecolor='black')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.title('Arabica Coffee Price Distribution')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.histplot(df['coffee_arabica'], kde=True)
plt.show()
```
