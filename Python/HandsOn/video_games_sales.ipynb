## EDA (Exploratory Data Analysis) vs Data Preprocessing

### EDA (Exploratory Data Analysis)

- Understand and explore the dataset 
  - → Examine the data to understand its structure, features, and overall quality.

- Detect Null Values 
  - → Identify missing values that may affect analysis or model performance.

- Detect Duplicates 
  - → Find repeated records that can bias results.

- Detect Outliers 
  - → Identify unusually high or low values that differ significantly from the rest of the data.

- Finding Trend / Pattern 
  - → Discover relationships, distributions, and patterns within the data.

- Visualization 
  - → Use charts and graphs to better understand and communicate the data.

### Data Preprocessing

- Clean and prepare the dataset 
  - → Convert raw data into a clean and usable format for analysis or machine learning.

- Handle Null Values 
  - → Remove or fill missing values using appropriate techniques.

- Handle Duplicates 
  - → Remove duplicate records to improve data quality.

- Handle Outliers 
  - → Treat or remove extreme values to reduce their impact on the model.

- Encoding (Changing Data Type) 
  - → Convert categorical/text data into numerical form so machine learning algorithms can process it.

## Detect Outliers

```python
for col in df.select_dtypes(include=['int64', 'float64']).columns:
    plt.figure(figsize=(3,2))
    plt.boxplot(df[col])
    plt.title(col)
    plt.show()
```

```python
cols = df.select_dtypes(include=['int64', 'float64']).columns

fig, axes = plt.subplots(3, 3, figsize=(10, 8))
axes = axes.flatten()

for i, col in enumerate(cols):
    axes[i].boxplot(df[col])
    axes[i].set_title(col)

plt.tight_layout()
plt.show()
```

```python
for col in df.select_dtypes(include='number'):
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    outliers = ((df[col] < q1 - 1.5 * iqr) | (df[col] > q3 + 1.5 * iqr)).sum()

    if outliers > 0:
        print(f"{col}: {outliers}")
```

## Remove Outliers

```python
sales_columns = ['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']

for col in sales_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    df = df[(df[col] >= lower) & (df[col] <= upper)]

print(df.shape)
```

```python
cols = df.select_dtypes(include=['int64', 'float64']).columns

fig, axes = plt.subplots(3, 3, figsize=(10, 8))
axes = axes.flatten()

for i, col in enumerate(cols):
    axes[i].boxplot(df[col])
    axes[i].set_title(col)

plt.tight_layout()
plt.show()
```

```python
for col in df.select_dtypes(include='number'):
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    outliers = ((df[col] < q1 - 1.5 * iqr) | (df[col] > q3 + 1.5 * iqr)).sum()

    if outliers > 0:
        print(f"{col}: {outliers}")
```

## Encoding

- The data still have str/object data types that is not good for analysis

- we need to encode

- Integer Column to Numeric Column

- ### Label Encoding
- ### Ordinal Encoding
- ### One hot Encoding

```python
# Example

data = {
    "Color" : ["Red", "Green", "Blue", "Red"],
    "Size" : ["Small", "Medium", "Large", "Large"],
    "Marital_Status":["Single", "Married", "Divorced", "Single"]
}
df1 = pd.DataFrame(data)
df1
```

```python
# Label Encoding -> It will convert the object column into numeric in an Alphabetical Order

from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()

df1["Color"] = le.fit_transform(df1["Color"])

df1
```

```python
# Ordinal Encoding -> Convert the object column into numeric in a order given by the user

from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=[["Small", "Medium", "Large"]])

df1["Size"] = oe.fit_transform(df1[["Size"]])

df1
```

```python
# One Hot Encoding ->

# df1 = pd.get_dummies(df1, columns=["Marital_Status"])

df1
```

```python
#  applying label encoding not good for marital_status
```

```python
# Apply Label encoding

df1["Marital_Status"] = le.fit_transform(df1["Marital_Status"])

df1
```
