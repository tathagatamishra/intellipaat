# PROBABILITY

Probability is simply a measure of how likely an event is to happen, expressed as a number between 0 and 1.

<img src="./images/probability-tree.png" width="380">

## Probability Terms and Definition
- Sample Space: The sample space is the set of all possible outcomes in a random experiment or event.
- Sample Point: A sample point represents a single outcome within a sample space, depicting a specific result of a random experiment or event.
- Experiment and Trails: An experiment or trial is a process with uncertain outcomes, often repeated to gather data or test hypotheses in scientific or mathematical contexts.
- Event: An event is a specific outcome or a collection of outcomes from a sample space in a random experiment or trial.
- Outcome: An outcome is a possible result or occurrence of a random experiment or event within the sample space.
- Complimentary event: A complementary event is the opposite of a given event, encompassing all outcomes not included in the given event.
- Impossible Event: An impossible event has zero probability of occurring, representing outcomes that cannot happen in a given situation or experiment.

## Types of Probabilities

1) **Marginal Prob**

2) **Joint Prob**

3) **Conditional Prob**

1) **Marginal Prob**

The probability of an event to happened without caring about any other event.

Ex: Prob of a person having covid 19, Probability of a person having asthma, Probability of a person renewing the subscription.

P(A) = Favorable Outcomes / Total outcomes

2) **Joint Prob**

What is prob of two or more events to happen at the same time.

Ex: Prob of a person to have covid and astha at the same time, Prob of rain and strom together, active user and renew the subscription, prob of virat and rohit to play the next wc and we win the next wc.
