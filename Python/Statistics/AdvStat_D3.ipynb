````python
import numpy as np
from scipy.stats import ttest_1samp, shapiro

# Simulated sample: 25 delivery times (in minutes)
# 1st delivery took 32 min, 2nd one took 35 min...
delivery_times = np.array([
    32, 35, 29, 38, 31, 34, 33, 40, 28, 36,
    37, 30, 39, 33, 35, 31, 42, 34, 36, 38,
    29, 33, 37, 35, 34])
````

### ttest_1samp function 
- can calculate sample mean, sample sd, root of ss
- cannot calculate population mean value

````python
#shapiro-wilk test for checking the normality of the sample
shapiro_stat, shapiro_p_value = shapiro(delivery_times)
print(shapiro_p_value)

#run 1-sample t-test on the data
t_statistic, p_value = ttest_1samp(delivery_times, popmean = 30)
print(p_value)
````

- almost 0 % possibility of null hypothesis to be true
