### Given a numpy array that has values = [1,2,3,4,5]. Perform the following operations.
- a. Insert the element 6 at the end of the array.
- b. Insert the element 10 at the beginning of the array.
- c. Insert the element 30, before 4.
- d. Update the element 5 as 50.
- e. Delete the element 50 in the updated array.
- f. Create a function to search for the element 30.
- g. Sort the array in ascending order.

```python
arr = np.array([1,2,3,4,5])
arr
```

```python
arr = np.append(arr, 6)
arr
```

```python
arr = np.insert(arr, 0, 10)
arr
```

```python
arr = np.insert(arr, 4, 30)
arr
```

```python
arr[arr == 5] = 50
arr
```

```python
arr = np.delete(arr, -2)
arr
```

```python
def search(arr, element):
    index = np.where(arr == element)[0]
    if index.size > 0:
        print(f"\nElement {element} found at index: {index[0]}")
    else:
        print(f"\nElement {element} not found in the array")

search(arr, 30)
```

```python
arr = np.sort(arr)
arr
```

### marks = np.array([45, 67, 89, 34, 90, 56])
- These are the marks of the students

- find highest, lowest, total and average marks
- add 5 grace marks to the marks of all students
- replace for all students who have got marks below 40 as Fail
- filter out students who have got marks more than 60.

```python
marks = np.array([45, 67, 89, 34, 90, 56])
marks
```

```python
print(np.max(marks))
print(np.min(marks))
print(np.sum(marks))
print(np.mean(marks))
```

```python
marks = marks + 5
marks
```

```python
result = marks.astype(object)
result[result < 40] = "Fail"
result
```

```python
filtered = marks[marks > 60]
filtered
```
