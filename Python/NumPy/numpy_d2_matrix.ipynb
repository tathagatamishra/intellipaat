```python
# Same shape — always works
a = np.array([[1, 2], [3, 4]])   # shape (2, 2)
b = np.array([[10, 20], [30, 40]]) # shape (2, 2)
print(a + b)  # [[11, 22], [33, 44]]

# Scalar broadcast — works with any shape
a = np.array([[1, 2], [3, 4]])   # shape (2, 2)
print(a + 10)  # [[11, 12], [13, 14]]

# (3, 3) + (3,) — trailing dims match, broadcasts row-wise
a = np.ones((3, 3))             # shape (3, 3)
b = np.array([1, 2, 3])        # shape (3,)
print(a + b)
# [[2., 3., 4.],
#  [2., 3., 4.],
#  [2., 3., 4.]]

# (3, 1) + (1, 3) — both broadcast to (3, 3)
a = np.array([[1], [2], [3]])   # shape (3, 1)
b = np.array([[10, 20, 30]])    # shape (1, 3)
print(a + b)
# [[11, 21, 31],
#  [12, 22, 32],
#  [13, 23, 33]]

# (3, 2) + (3,) with mismatch — FAILS
a = np.ones((3, 2))            # shape (3, 2)
b = np.array([1, 2, 3])        # shape (3,)  — trailing dim 3 ≠ 2
# ValueError: operands could not be broadcast together with shapes (3,2) (3,)
```

### Transpose

```python
# rows and columns get interchanged with each other

# rows = 4
# columns = 2

a = np.array([[1,2], [3,4], [5,6], [7,8]])
a
```

```python
np.transpose(a)
```

### Flatten

```python
# converting the data  into 1D

b = a.flatten()
b
```

```python
# similar to flatten

# diff --> flatten creates a copy of the data but ravel creates view of the data
```

```python
a
```
