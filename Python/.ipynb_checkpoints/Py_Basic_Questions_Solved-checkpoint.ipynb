```python
# QUESTION-1
# Write a program to evaluate employee salaries.

def salary_category(salary):
    if salary >= 50000:
        return "High"
    elif salary >= 30000:
        return "Medium"
    else:
        return "Low"

salaries = []
for i in range(1, 6):
    salary = int(input(f"Enter salary of Employee {i}: "))
    salaries.append(salary)

counts = {"High": 0, "Medium": 0, "Low": 0}

for i, salary in enumerate(salaries, 1):
    category = salary_category(salary)
    counts[category] += 1
    print(f"Employee {i}: {category}")

print("\nCategory Summary:")
for cat, count in counts.items():
    print(f"  {cat}: {count} employee(s)")
```

```python
# QUESTION-2
# Number analysis without built-in reverse methods

number = int(input("Enter a number: "))

# 1. Palindrome check using loop
reversed_num = 0
temp = number
while temp > 0:
    digit = temp % 10
    reversed_num = reversed_num * 10 + digit
    temp //= 10

print("Palindrome    :", "Yes" if number == reversed_num else "No")

# 2. Even or Odd
print("Even/Odd      :", "Even" if number % 2 == 0 else "Odd")

# 3. Count digits using loop
count = 0
temp = number
while temp > 0:
    count += 1
    temp //= 10
print("Total Digits  :", count)

# 4. Sum of digits using loop
digit_sum = 0
temp = number
while temp > 0:
    digit_sum += temp % 10
    temp //= 10
print("Sum of Digits :", digit_sum)
```

```python
# QUESTION-3
# List operations using loops and conditions

li = [10, 20, 10, 30, 40, 20, 50, 30]
print("Original list :", li)

# 1. Remove duplicates manually (no set)
unique = []
for item in li:
    if item not in unique:
        unique.append(item)
print("After removing duplicates:", unique)

# 2. Remove elements divisible by 10
filtered = []
for item in unique:
    if item % 10 != 0:
        filtered.append(item)
print("After removing multiples of 10:", filtered)

# 3. Sort in descending order (bubble sort)
for i in range(len(filtered)):
    for j in range(0, len(filtered) - i - 1):
        if filtered[j] < filtered[j + 1]:
            filtered[j], filtered[j + 1] = filtered[j + 1], filtered[j]

print("Final list (descending):", filtered)
```

```python
# QUESTION-4
# Login system with max 3 attempts

users = {
    "admin": "1234",
    "user1": "pass1",
    "user2": "pass2"
}

attempts = 0
max_attempts = 3

while attempts < max_attempts:
    username = input("Enter username: ")
    password = input("Enter password: ")
    attempts += 1

    if username not in users:
        print("Username not found")
    elif users[username] != password:
        print("Wrong Password")
    else:
        print("Login Successful")
        break

    if attempts == max_attempts:
        print("Account Locked")
```

```python
# QUESTION-5
# lambda, map, filter, sorted — filter → map → sorted

nums = [12, 5, 67, 34, 89, 2, 44]
print("Original list      :", nums)

# 1. Filter even numbers
evens = list(filter(lambda x: x % 2 == 0, nums))
print("Even numbers       :", evens)

# 2. Square each even number
squared = list(map(lambda x: x ** 2, evens))
print("Squared            :", squared)

# 3. Sort by last digit
final = sorted(squared, key=lambda x: x % 10)
print("Sorted (last digit):", final)
```
