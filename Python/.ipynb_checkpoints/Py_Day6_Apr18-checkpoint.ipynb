```python
def add(a,b): #a and b are the parameters
  print(a+b)

add(10,20)  #arguments- the actual values passed into the parameters
add(100,90)

#Why parameters are required?
#They make the functions reusable with different values
```

```python
def sumup(a,b):
    # a=input("Input first number:")
    a=int(a)
    # b=input("Input second number:")
    b=int(b)
    print(a+b)
#calling UDF sumup
a=2
b=3
sumup(a,b)
```

```python
def sumup(a,b):
    a=input("Input first number:")
    a=int(a)
    b=input("Input second number:")
    b=int(b)
    print(a+b)
    
#calling UDF sumup
sumup(0,0)
```

```python
def totalMarks(m1, m2, m3):
    total = m1 + m2 + m2
    return total

def passORfail(total):
    if total > 150:
        return "Pass"
    else:
        return "Fail"

passORfail(totalMarks(50, 67, 82))
```

```python
# write a code the subject marks should be min of 35 and cumulative total should be 150 for passing the exams?
def result(m1,m2,m3):
  total=m1+m2+m3

  if m1>=35 and m2>=35 and m3>=35 and total>=150:
    print('Pass')
  else:
    print('Fail')

m1=int(input())
m2=int(input())
m3=int(input())

result(m1,m2,m3)
```

```python
even_odd = lambda x: "Even" if x % 2 == 0 else "Odd"
num = int(input("Enter a number: "))
print(even_odd(num))
```
