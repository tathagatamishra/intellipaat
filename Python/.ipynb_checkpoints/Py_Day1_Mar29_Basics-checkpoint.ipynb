```python
# ------- PYTHON BASICS - Day 1 -------

# Python is a high level programming language

# Close to English that is human understandable

# Python first Compiles the Source Code into Bytecode

# Then Bytecode get Interpreted

# Source code [print("Hello")] --> Byte code --> Machine Code
```

```python
# Variables & Data Types

# Text
game = "PUBG"

# Whole number
kills = 25

# Decimal number
kd = 8.5

# True/False
headshotOnly = True

# Print everything
print("Game:", game)
print("Kills:", kills)
print("KD:", kd)
print("Headshot only mode:", headshotOnly)

# Check what type each variable is
print(type(game))
print(type(kills))
print(type(kd))
print(type(headshotOnly))
```
