```python
#create a class stu
# __init__ -> name ,marks
#Create methods->
# a-display() -> print name and marks
# b-is_pass() -> it will check if marks >=40 pass else fail

#create an obj for the class and call both the methods

class Student:
    def __init__(this, name, marks):
        this.name = name
        this.marks = marks
        
        def display(this):
            print(this.name + ": " + str(this.marks))
            
        def isPass(this):
            if this.marks >= 40:
                print("Pass")
            else:
                print("Fail")
                
s1 = Student('Bob', 67)
s1.display()
s1.isPass()

# It has the indentation issue
```

```python
# faced indentation issue, but fixed it, 
# initially i put the display and isPass inside __init__, 
# but later I undertood and fixed it

class Student:
    def __init__(this, name, marks):
        this.name = name
        this.marks = marks

    def display(this):
        print(this.name + ": " + str(this.marks))

    def isPass(this):
        if this.marks >= 40:
            print("Pass")
        else:
            print("Fail")

s1 = Student('Bob', 67)
s1.display()
s1.isPass()
```

## INHERITANCE

```python
# Inheritance allows one class to acquire the properties and behavior of another class

#in simple terms-> a new class can reuse code from an existing class

#Parent class
#Child class
#Child class inherits from parent class

class Vehicle: #parent class
  def start():
    print("Vehicle started")

class Car(Vehicle): # in the () comes the name of the parent class
  def drive():
    print("Car can drive")

Car.start()
Car.drive()

# class name also work like object
```

```python
# if use self, the have to create a var

class Vehicle: #parent class
  def start(self):
    print("Vehicle started")

class Car(Vehicle): # in the () comes the name of the parent class
  def drive(self):
    print("Car can drive")

c = Car()
c.start()
c.drive()
```

```python
#Types of inheritance:-
# 1-Single
# 2- Multiple
# 3- Multi-level
# 4- Hierarchial
```

```python
#Single inheritance- One child class inherits from parent class
class Animal:
  def eat(self):
    print("Animal eats food")

class Dog(Animal):
  def bark(self):
    print("Dog barks")

d=Dog()
d.eat()
```

```python
#Multiple Inheritance-> One child class inherits from multiple parents

class Father:
  def skill1(self):
    print("Driving")

class Mother:
  def skill2(self):
    print("Cooking")

class Child(Father, Mother):
  pass

c=Child()
c.skill1()
c.skill2()
```

```python
#What if both parents have same method name

class Father:
  def show(self):
    print("Father")


class Mother:
  def show(self):
    print("Mother")

# class Child(Mother,Father): same thing
class Child(Father,Mother):   #internal search order-> Child->Father->Mother
  def show(self):
    print('Child')

c=Child()
c.show()
print(Child.mro())

# C3 linearization is a deterministic algorithm used in Python

#Python follows Method Resolution Order
#it checks for methods from left to right

#current class->Parent classes(left to right)->Their parents
```

```python
#Super()-> Calls the next class in MRO, not neccessarily the parent
#super()-> doesn't mean
class Father:
  def show(self):
    print("Father")


class Mother:
  def show(self):
    print("Mother")

class Child(Father,Mother):
  def show(self):
    print('Child')
    super().show()

c=Child()
c.show()
```

```python
class Father:
  def show(self):
    print("Father")
    super().show()

class Mother:
  def show(self):
    print("Mother")

class Child(Father,Mother):
  def show(self):
    print('Child')
    super().show()

c=Child()
c.show()
```

```python
#Multilevel Inheritance-
# It means a class inheritsfrom another class and then another class inherits from that derived class

#Grandparent->Parent->Child

class Grandparent:
  def show_gp(self):
    print("I am Grandparent")

class Parent(Grandparent):
  def show_p(self):
    print("I am Parent")

class Child(Parent):
  def show(self):
    print("I am Child")

c=Child()
c.show_gp()
```

```python
#Hierarchial Inheritance:-
#means multiple child classes inherit from the same parent class

#one parent-> many children

class Animal:
  def eat(self):
    print("animal eats")

class Dog(Animal):
  def bark(self):
    print("dog barks")

class Cat(Animal):
  def meows(self):
    print("cat meows")

d=Dog()
c=Cat()

d.eat()
d.bark()

c.eat()
c.meows()
```

```python
#Hybrid inheritance-> Cobination of 2 or more types of inheritance

class Person:
  def details(self):
    print('Person')

class Student(Person):   #Hierarchial
  def study(self):
    print("Studying")

class Emp(Person):
  def work(self):
    print('work')

class Intern(Student,Emp):    #Multipl
  def role(self):
    print("I am an intern")


obj=Intern()
obj.details()
obj.study()
```

```python
class Student:
  def __init__(self,name,marks):
    self.name=name
    self.marks=marks

students=[]

students.append(Student('anurag',78))
students.append(Student('tathagata',89))
students.append(Student('garima',70))
students.append(Student('aman',98))
students.append(Student('abc',87))

#fetch and print names
names=[]
for s in students:
  names.append(s.name)

#print all names
for name in names:
  print(name, end=' ')
```
