```python
#Class -> A class is a blueprint or template

# class=blueprint of a house

class Student:
  def __init__(self,name,age):  #init is special function . It runs automatically when we create an object
    self.name=name           #name and age are temporary variables (input variables)
    self.age=age             #self.name and self.age = these are stored inside the object

s1=Student('Karan',27)
s1.name

s2=Student('Monika',36)
s2.name
```

```python
#create a class stu
# __init__ -> name ,marks
#Create methods->
# a-display() -> print name and marks
# b-is_pass() -> it will check if marks >=40 pass else fail

#create an obj for the class and call both the methods

class Student:
    def __init__(this, name, marks):
        this.name = name
        this.marks = marks
        
        def display(this):
            print(this.name + ": " + str(this.marks))
            
        def isPass(this):
            if this.marks >= 40:
                print("Pass")
            else:
                print("Fail")
                
s1 = Student('Bob', 67)
s1.display()
s1.isPass()

# It has the indentation issue
```

```python
# faced indentation issue, but fixed it, 
# initially i put the display and isPass inside __init__, 
# but later I undertood and fixed it

class Student:
    def __init__(this, name, marks):
        this.name = name
        this.marks = marks

    def display(this):
        print(this.name + ": " + str(this.marks))

    def isPass(this):
        if this.marks >= 40:
            print("Pass")
        else:
            print("Fail")

s1 = Student('Bob', 67)
s1.display()
s1.isPass()
```
