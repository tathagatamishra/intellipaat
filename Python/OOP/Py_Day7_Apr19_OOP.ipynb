## INHERITANCE

```python
# Inheritance allows one class to acquire the properties and behavior of another class

#in simple terms-> a new class can reuse code from an existing class

#Parent class
#Child class
#Child class inherits from parent class

class Vehicle: #parent class
  def start():
    print("Vehicle started")

class Car(Vehicle): # in the () comes the name of the parent class
  def drive():
    print("Car can drive")

Car.start()
Car.drive()

# class name also work like object
```

```python
# if use self, the have to create a var

class Vehicle: #parent class
  def start(self):
    print("Vehicle started")

class Car(Vehicle): # in the () comes the name of the parent class
  def drive(self):
    print("Car can drive")

c = Car()
c.start()
c.drive()
```

```python
#Types of inheritance:-
# 1-Single
# 2- Multiple
# 3- Multi-level
# 4- Hierarchial
```

```python
#Single inheritance- One child class inherits from parent class
class Animal:
  def eat(self):
    print("Animal eats food")

class Dog(Animal):
  def bark(self):
    print("Dog barks")

d=Dog()
d.eat()
```

```python
#Multiple Inheritance-> One child class inherits from multiple parents

class Father:
  def skill1(self):
    print("Driving")

class Mother:
  def skill2(self):
    print("Cooking")

class Child(Father, Mother):
  pass

c=Child()
c.skill1()
c.skill2()
```

```python
#What if both parents have same method name

class Father:
  def show(self):
    print("Father")


class Mother:
  def show(self):
    print("Mother")

# class Child(Mother,Father): same thing
class Child(Father,Mother):   #internal search order-> Child->Father->Mother
  def show(self):
    print('Child')

c=Child()
c.show()
print(Child.mro())

# C3 linearization is a deterministic algorithm used in Python

#Python follows Method Resolution Order
#it checks for methods from left to right

#current class->Parent classes(left to right)->Their parents
```

```python
#Super()-> Calls the next class in MRO, not neccessarily the parent
#super()-> doesn't mean
class Father:
  def show(self):
    print("Father")


class Mother:
  def show(self):
    print("Mother")

class Child(Father,Mother):
  def show(self):
    print('Child')
    super().show()

c=Child()
c.show()
```

```python
class Father:
  def show(self):
    print("Father")
    super().show()

class Mother:
  def show(self):
    print("Mother")

class Child(Father,Mother):
  def show(self):
    print('Child')
    super().show()

c=Child()
c.show()
```

```python
#Multilevel Inheritance-
# It means a class inheritsfrom another class and then another class inherits from that derived class

#Grandparent->Parent->Child

class Grandparent:
  def show_gp(self):
    print("I am Grandparent")

class Parent(Grandparent):
  def show_p(self):
    print("I am Parent")

class Child(Parent):
  def show(self):
    print("I am Child")

c=Child()
c.show_gp()
```

```python
#Hierarchial Inheritance:-
#means multiple child classes inherit from the same parent class

#one parent-> many children

class Animal:
  def eat(self):
    print("animal eats")

class Dog(Animal):
  def bark(self):
    print("dog barks")

class Cat(Animal):
  def meows(self):
    print("cat meows")

d=Dog()
c=Cat()

d.eat()
d.bark()

c.eat()
c.meows()
```

```python
#Hybrid inheritance-> Cobination of 2 or more types of inheritance

class Person:
  def details(self):
    print('Person')

class Student(Person):   #Hierarchial
  def study(self):
    print("Studying")

class Emp(Person):
  def work(self):
    print('work')

class Intern(Student,Emp):    #Multipl
  def role(self):
    print("I am an intern")


obj=Intern()
obj.details()
obj.study()
```

## Encapsulation

```python
#Encapsulation -> wrapping data(variables) and methods(functions) together inside a class and restrivcting direct access to some of them.

#key-ideas:-
# a-Control access
# b-Protect data
# c- hide internal details
```

```python
# public
class Student:
    def __init__(self, name, age, sex):
        self.name = name
        self.age = age
        self.sex = sex

s1 = Student('Bob', 20, 'M')
print(s1.name)
print(s1.age)
print(s1.sex)
```

```python
# protected
class Student:
    def __init__(self, name, age, sex):
        self._name = name
        self._age = age
        self._sex = sex

s1 = Student('Bob', 20, 'M')
print(s1._name)
print(s1._age)
print(s1._sex)
```

```python
class BankAccount:
    def __init__(self, initialBalance=0):
        self.__balance = initialBalance

    def deposit(self, amount):
        if amount <= 0:
            print("amount must be positive")
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            print("amount must be positive")
        if amount > self.__balance:
            print("you are broke go rob a bank")
        self.__balance -= amount

    def getBalance(self):
        return self.__balance


acc = BankAccount(200)

acc.deposit(100)
acc.withdraw(1000)

print(acc.getBalance())   
```

```python
class Shape:
    def area(self):
        pass
        
class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        pi = 3.1416
        return pi * self.radius * self.radius

rect = Rectangle(10, 5)
circle = Circle(7)

print(rect.area())
print(circle.area())
```

```python
from abc import ABC, abstractmethod

class RemoteControl(ABC):

    @abstractmethod
    def power_on(self):
        pass

    @abstractmethod
    def power_off(self):
        pass

    def mute(self):
        print("Muted")


class TVRemote(RemoteControl):

    def power_on(self):
        print("TV ON")

    def power_off(self):
        print("TV OFF")

    def change_channel(self, channel):
        print(f"Channel changed to {channel}")

    def adjust_volume(self, level):
        print(f"Volume set to {level}")


tv = TVRemote()
tv.power_on()
tv.change_channel('BBC')
tv.adjust_volume(150)
tv.mute()
tv.power_off()
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} (₹{self.price})"


class Cart:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)
        print(f"'{product.name}' added to cart")

    def remove_product(self, name):
        for product in self.products:
            if product.name.lower() == name.lower():
                self.products.remove(product)
                print(f"'{product.name}' removed from cart")

    def total_price(self):
        total = sum(p.price for p in self.products)
        for p in self.products:
            print(f"  {p}")
        print(f"Total: ₹{total}")


p1 = Product("R6", 600)
p2 = Product("DBD", 1000)
p3 = Product("COD", 2000)
p4 = Product("GTA", 1200)

cart = Cart()
cart.add_product(p1)
cart.add_product(p2)
cart.add_product(p3)
cart.add_product(p4)
print("-------------")
cart.remove_product("COD")
print("-------------")
cart.total_price()
```

```python
class Camera:
    def click(self):
        print("Photo captured!")

class Phone:
    def call(self):
        print("Calling...")

class SmartPhone(Camera, Phone):
    def click(self): 
        print("High-res photo captured!")

    def call(self):               # overrides Phone's call
        print("📱 SmartPhone Call — HD voice call connected!")

    def browse(self):             # SmartPhone's own method
        print("🌐 Browsing the internet...")


# Create object and call all methods
sp = SmartPhone()
sp.click()
sp.call()
sp.browse()
```

```python
class Bus:
    def __init__(self, total_seats):
        self.total_seats = total_seats
        self.booked_seats = 0             # starts at 0

    def book_seat(self, seats=1):
        if seats <= 0:
            print("Invalid number of seats.")
        elif seats > self.available_seats():
            print(f"Cannot book {seats} seat(s). Only {self.available_seats()} available.")
        else:
            self.booked_seats += seats
            print(f"✅ {seats} seat(s) booked. Booked: {self.booked_seats} | Available: {self.available_seats()}")

    def cancel_seat(self, seats=1):
        if seats <= 0:
            print("Invalid number of seats.")
        elif seats > self.booked_seats:
            print(f"Cannot cancel {seats} seat(s). Only {self.booked_seats} are booked.")
        else:
            self.booked_seats -= seats
            print(f"❌ {seats} seat(s) cancelled. Booked: {self.booked_seats} | Available: {self.available_seats()}")

    def available_seats(self):
        return self.total_seats - self.booked_seats

    def status(self):
        print(f"Total: {self.total_seats} | Booked: {self.booked_seats} | Available: {self.available_seats()}")


# ── Demo ──
bus = Bus(total_seats=10)
bus.status()

bus.book_seat(4)     # ✅ booked
bus.book_seat(3)     # ✅ booked
bus.book_seat(5)     # ❌ exceeds available

bus.cancel_seat(2)   # ✅ cancelled
bus.cancel_seat(10)  # ❌ exceeds booked

bus.status()
```
