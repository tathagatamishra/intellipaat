```python
#Object oriented programming is a programming paradigm where programs are designed using objects and classes
#This makes the programs:-
# a- Easier to understand
# b- Easier to manage
# c- Easier to reuse
# d- Easier to scale for large systems

#Why do we need OOP?
# Imagine we are creating a website without OOP:-
# a- diff to maintain
# b- diff to debug
# c- diff to expand

# OOP helps us to organize code logically, similar to how things exist in real world
```

```python
#Class -> A class is a blueprint or template

# class=blueprint of a house

class Student:
  def __init__(self,name,age):  #init is special function . It runs automatically when we create an object
    self.name=name           #name and age are temporary variables (input variables)
    self.age=age             #self.name and self.age = these are stored inside the object

s1=Student('Karan',27)
s1.name

s2=Student('Monika',36)
s2.name
```
