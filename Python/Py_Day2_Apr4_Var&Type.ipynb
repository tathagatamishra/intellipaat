## Variable

- Variables are containers for storing data values
- Python has no command for declaring a variable.
- A variable is created the moment you first assign a value to it.

```python
# think it like math, suppose 
x = 2 
y = 1 
x + y
# we could have write 2 + 1 directly
# but we create variable x and y  to store the numbers to perform addition
```

```python
a = 2

del a       # for deleting variable
```

### Rules for creating variable

1. A variable can start with an alphabet or and underscore
2. Can't start with a number, but it can have number in between or in the end
3. We can't have special characters in variable (@#$%^&*)
4. Can't have spaces between
5. Variables are case sensitive (name & Name is different)
6. Cant't use keywords (for, while, if, print)

### Data types

> datatypes specify the type of data that a variable can store


```python
# Variables & Data Types

# Text
game = "PUBG"

# Whole number
kills = 25

# Decimal number
kd = 8.5

# True/False
headshotOnly = True


# Check what type each variable is
print(type(game))
print(type(kills))
print(type(kd))
print(type(headshotOnly))
```

```python
# integer
a = 20
print(type(a))


# float     ->  any numeric value positive or negative with a decimal
b = 0.0
print(type(b))

c = -67.
print(type(c))


# string    -> can store a letter, word and a sentence
d = "Slim Shady"
print(type(d))
```

```python
# We can not have object name as variable

# but we can use object as value of a variable

a = print
a("Hi")     # now a works as print

# object and keyword
print()     # it is an object
# for()       # is an keyword
```

```python
# Complex Number

# used to store real and imaginary numbers

# j is the only imaginary number
x = 2 + 3j
print(type(x))

y = 2j + 3j # but here j will have same value
# y = 2j + 3k # we can do this
# so we cant have multiple imaginary number in a single variable
y = 2j + 3J
# seems like both j & J are reserved for complex number
```

```python
# int to str

a = 100
a = str(a)
print(a, type(a))

# if you multiply number with a string --> it just print the string multiple times
a*5
```

```python
# str to int

b = '100'
b = int(b)
print(b, type(b))

# for converting any str to int you need to have an int datatype inside quotes " "
c = '10.0'
c = int(c)
print(c, type(c))
# error
```

```python
# string or int to float
a = "19"
a = float(a)
print(a, type(a))
```

```python
# bool to int
a = True
a = int(a)
print(a, type(a))

b = False
b = int(b)
print(b, type(b))
```

```python
# int to bool

# any non zero number is considered True
# the number 0, 0.0, -0 is False

a = -2
a = bool(a)
print(a, type(a))

b = 2
b = bool(b)
print(b, type(b))

c = -0
c = str(c)
print(c, type(c))
```

# OPERATORS

- are symbols or keywords that performs operations on variables or values/operands

5           +           3
operand  operator  operand

```python
# Arithmetic operators
# +  -  *  /  %  **  //

a = 5
b = 3
print(a + b)    # addition
print(a - b)    # subtraction
print(a * b)    # multiplication
print(a / b)    # quotient
print(a % b)    # remainder
print(a ** b)   # exponentiation
print(a // b)   # floor division
```

```python
print(True + False + True + True + True - False - False)
print(1 + 0 + 1 + 1 + 1 - 0 - 0)
```
