### Loc and iLoc

```python
# loc -- location based indexes or custom indexes
# iloc -- always works upon the basis of integer
```

```python
import pandas as pd
```

```python
s1 = pd.Series([1,2,3,4,5])
s1
```

```python
print(s1.loc[0])
```

```python
print(s1.iloc[0])
```

```python
s2 = pd.Series([1,2,3,4,5], index = ['a','b','c','d','e'])
s2
```

```python
print(s2.loc['a'])
# print(s2.loc[0]) -- wont work
```

```python
# print(s2.iloc['a']) -- wont work
print(s2.iloc[0])
```

```python
s3 = pd.Series([1,2,3,4,5], index = [5,6,7,8,9])
s3
```

```python
print(s3.loc[5])
print(s3.iloc[0])
```

### Slicing

```python
s3.iloc[0:2] # last index is always exclusive
```

```python
s3.loc[5:15] # only with loc the last index is included
```

```python
s2 = pd.Series([1,2,3,4,5], index = ['a','b','c','d','e'])
s2
# using loc print the values from 1 to 4
```

```python
s2.loc['a':'d']
```

```python
s2.iloc[0:4]
```

```python
s3 = pd.Series([1,2,3,4,5], index = [5,6,7,8,9])
s3
```

```python
new = pd.Series([6], index=[0])
s3 = pd.concat([s3, new])
s3
```

```python
s0 = pd.Series([1,2,3,4,5])
s0
```

```python
s0.loc['new_index'] = 10
s0
```

```python
s0['cool'] = 67
s0
```
