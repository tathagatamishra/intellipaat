```python
#create a series which incudes name, age, degination, current year
```

```python
myData = pd.Series(
    ["Tatha", 102, "Software Engineer", 2026],
    index=["Name", "Age", "Designation", "Current Year"]
)
myData
```

```python
# Series vs Array
```

```python
my_tuple = ("GG", 67, "Engineer", 2026)

series_tuple = pd.Series(my_tuple)
```

```python
my_set = {10, 20, 30, 40, 50}

series_set = pd.Series(list(my_set))

series_set
```

```python
sales_dict = {
    "D1": 15000,
    "D2": 22000,
    "D3": 18500,
    "D4": 30000,
    "D5": 27500
}

sales_dict
```

```python
sales_series = pd.Series(sales_dict)

sales_series
```

```python
sales_series.name = 'Sales'
```

```python
sales_series
```
