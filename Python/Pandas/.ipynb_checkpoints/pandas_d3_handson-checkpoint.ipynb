5) Find if there are any null values.
6) Drop the null values
7) Create a function which takes the data frame( as a parameter) and the function should fill the null values in the string/object data type column with mode of the column and the numerical col null values with the mean.

```python
# 5) Find if there are any null values
print("NULL VALUES PER COLUMN:")
print(df.isnull().sum())

print("\nANY NULL VALUES IN DATAFRAME:", df.isnull().values.any())
print("TOTAL NULL VALUES:", df.isnull().sum().sum())
```

```python
# 6) Drop null values
df_dropped = df.dropna()
print("\nShape BEFORE dropping nulls:", df.shape)
print("Shape AFTER  dropping nulls:", df_dropped.shape)
```

```python
# 7) Function to fill nulls — mode for object cols, mean for numeric cols
def fill_null_values(df):
    df = df.copy()  # avoid modifying the original
    for col in df.columns:
        if df[col].isnull().sum() > 0:                  # only process cols with nulls
            if df[col].dtype == 'object':
                df[col] = df[col].fillna(df[col].mode()[0])   # mode for string cols
            else:
                df[col] = df[col].fillna(df[col].mean())      # mean for numeric cols
    return df

df_filled = fill_null_values(df)

print("\nNull values BEFORE filling:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nNull values AFTER filling:")
print(df_filled.isnull().sum().sum(), "← should be 0")
```
